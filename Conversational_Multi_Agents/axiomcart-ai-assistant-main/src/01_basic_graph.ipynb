**Goal:** Show the core LangGraph mental model: state moves through nodes.

```python
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from langchain_core.messages import AnyMessage

class CartState(TypedDict): #typed dict
    request: str # can't int
    response: str
    boolean_flag: bool
    message_history: list[AnyMessage]

def understand_request(state: CartState):
    request = state["request"].lower()
    category = "support" if "order" in request else "product"
    return {"response": f"Detected category: {category}"}

builder = StateGraph(CartState)
builder.add_node("understand_request", understand_request)
builder.add_edge(START, "understand_request")
builder.add_edge("understand_request", END)

graph = builder.compile()
graph.invoke({"request": "Where is my order?", "response": ""})
```

### Takeaway

> “A node is just a function. State goes in, the node returns an update, and LangGraph carries the updated state forward.”
